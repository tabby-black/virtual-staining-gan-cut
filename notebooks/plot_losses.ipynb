```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# plot generator and discriminator losses
# Avoid log(0) by clipping values
x = np.linspace(1e-5, 1 - 1e-5, 500)

# BCE-based GAN losses
generator_loss = -np.log(x)          # -log(D(G(x)))
discriminator_loss = -np.log(1 - x)  # -log(1 - D(G(x)))

plt.figure(figsize=(6, 4))

plt.plot(x, generator_loss, label="Generator Loss")
plt.plot(x, discriminator_loss, label="Discriminator Loss")

plt.xlabel("Discriminator output D(G(x))")
plt.ylabel("Loss")
plt.title("Adversarial Loss vs Discriminator Confidence")

plt.legend()
plt.grid()

plt.show()
```
