```python
import numpy as np
from spectral import open_image, envi
import matplotlib.pyplot as plt
```

```python

# I can change this to add gaussian noise with probability p if needed
def add_gaussian_noise(image, sigma=0.01):
    """
    image: numpy array, assumed float32/float64 in [0, 1]
    # this is true because of preprocessing
    sigma: standard deviation of Gaussian noise

    returns: noisy image clipped to [0, 1]
    """
    # noise ~ N(0, sigma^2)
    noise = np.random.normal(loc=0.0, scale=sigma, size=image.shape)
    noisy = image + noise
    noisy = np.clip(noisy, 0.0, 1.0)
    return noisy
```

```python
hdr_path = "../datasets/histology/trainA/P9_ROI_02_C10_T_raw_preprocessed_bottom_right.hdr"
image = open_image(hdr_path).load()

noisy = add_gaussian_noise(image)

# have to visualise by looking at one HSI band because we cannot display all at once
band = 10

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(image[:, :, band], cmap='gray')
plt.title("Original HSI band")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(noisy[:, :, band], cmap='gray')
plt.title("Noisy HSI band")
plt.axis("off")

plt.show()
```
