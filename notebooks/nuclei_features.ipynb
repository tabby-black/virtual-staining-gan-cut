```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# at the moment this saves nuclei counts and image names to a csv in histology_analysis
def extract_features_to_csv(mask_dir):

    image_names = []
    # array of nuclei count of each real RGB test image
    nuclei_counts = []

    i = 0
    for filename in os.listdir(mask_dir):
        i += 1
        mask_path = os.path.join(mask_dir, filename)
        mask = np.load(mask_path)
    
        count = mask.max()
    
        image_names.append(filename.replace("_mask.npy", ""))
        # add nuclei count to list
        nuclei_counts.append(count)
        print("Nuclei count in image", i, "added to list!")

    #nuclei_counts = np.array(nuclei_counts)
    print(nuclei_counts)

    # save nuclei counts so don't have to recompute each time
    df = pd.DataFrame({
        "image_name": image_names,
        "nuclei_count": nuclei_counts
    })

    # csv file = real_RGB_nuclei_counts.csv
    df.to_csv("/results/histology_analysis/real_RGB_nuclei_counts.csv", index=False)
```

```python
# load data from a csv file and plot line graph
# first plot real RGB frequency against nuclei count 
def read_csv_plot_linegraph(csv_path):
    df = pd.read_csv(csv_path)
    # nuclei counts is restored to an array from the csv
    counts = df["nuclei_count"].values

    unique_counts, frequencies = np.unique(counts, return_counts=True)

    plt.figure(figsize=(6, 4))
    plt.plot(unique_counts, frequencies)

    plt.xlabel("Nuclei count per image")
    plt.ylabel("Frequency")
    plt.title("Nuclei count frequency distribution")

    plt.show()
```

```python
# load data from a csv file and plot histogram
# first plot real RGB frequency against nuclei count
def read_csv_plot_histogram(csv_path):
    df = pd.read_csv(csv_path)
    # nuclei counts is restored to an array from the csv
    counts = df["nuclei_count"].values
    plt.figure(figsize=(6, 4))
    plt.hist(counts, bins=10)

    plt.xlabel("Nuclei count per image")
    plt.ylabel("Frequency")
    plt.title("Distribution of nuclei count (real images)")

    plt.show()
```
