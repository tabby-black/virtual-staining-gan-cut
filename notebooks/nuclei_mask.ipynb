```python
import os
import numpy as np
import matplotlib.pyplot as plt
from cellpose import models, io
from cellpose.plot import mask_overlay
```

```python
real_path = "./results/hsi_to_rgb_cyclegan_rep2/test_20/images/real_B/P8_ROI_02_C15_NT_raw_preprocessed_7.png"
fake_path = "./results/hsi_to_rgb_cyclegan_rep2/test_20/images/fake_B/P8_ROI_02_C15_NT_raw_preprocessed_7.png"

model = models.CellposeModel(model_type="nuclei", gpu=True)

for label, image_path in [("real_B", real_path), ("fake_B", fake_path)]:
    img = io.imread(image_path)

    masks, flows, styles = model.eval(
        img,
        channels=[0, 0],   # try [3, 0] if nuclei are blue/purple
        diameter=None
    )

    overlay = mask_overlay(img, masks)

    print(f"{label}: nuclei detected =", masks.max())

    plt.figure(figsize=(15, 5))

    plt.subplot(1, 3, 1)
    plt.imshow(img)
    plt.title(f"{label} original")
    plt.axis("off")

    plt.subplot(1, 3, 2)
    plt.imshow(masks, cmap="nipy_spectral")
    plt.title(f"{label} labelled mask")
    plt.axis("off")

    plt.subplot(1, 3, 3)
    plt.imshow(overlay)
    plt.title(f"{label} overlay")
    plt.axis("off")

    plt.show()
```
