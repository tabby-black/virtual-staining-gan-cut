```python
from spectral import open_image, envi
import numpy as np
import os
import glob
import matplotlib.pyplot as plt

from PIL import Image
#from itertools import product
```

```python
def patch_hsi(image, i=0):
    
    H, W, B = image.shape

    # compute 4 non-overlapping patches
    # I need to make patch width 500 rather than 502 so it is divisible by 4 for generator
    # save half heights and widths that are divisble by 4
    half_h = (H // 2) // 4 * 4
    half_w = (W // 2) // 4 * 4
    patches = {
        "top_left": image[0:half_h,   0:half_w, :],
        "top_right": image[0:half_h,   W-half_w:W, :],
        "bottom_left": image[H-half_h:H,   0:half_w, :],
        "bottom_right": image[H-half_h:H,   W-half_w:W, :]
    }

    return patches
```

```python
# patch an example hsi image
hdr_path = "../datasets/histology/testA/P12_ROI_01_C10_T_raw_preprocessed.hdr"
image = open_image(hdr_path).load()
patch_dict = patch_hsi(image)

band = 10

plt.figure(figsize=(15, 4))

plt.subplot(1, 5, 1)
plt.imshow(image[:, :, band], cmap='gray')
plt.title("Original")
plt.axis("off")

plt.subplot(1, 5, 2)
plt.imshow(patch_dict["top_left"][:, :, band], cmap='gray')
plt.title("Top left")
plt.axis("off")

plt.subplot(1, 5, 3)
plt.imshow(patch_dict["top_right"][:, :, band], cmap='gray')
plt.title("Top right")
plt.axis("off")

plt.subplot(1, 5, 4)
plt.imshow(patch_dict["bottom_left"][:, :, band], cmap='gray')
plt.title("Bottom left")
plt.axis("off")

plt.subplot(1, 5, 5)
plt.imshow(patch_dict["bottom_right"][:, :, band], cmap='gray')
plt.title("Bottom right")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
def patch_rgb(rgb, i=0):

    W, H = rgb.size

    # compute 4 non-overlapping patches
    # safe half heights and half widths that are divisible by 4
    half_h = (H // 2) // 4 * 4
    half_w = (W // 2) // 4 * 4
    patches = {
        "top_left": rgb.crop((0,    0,  half_w,   half_h)),
        "top_right": rgb.crop((W-half_w, 0,  W,  half_h)),
        "bottom_left": rgb.crop((0, H-half_h,   half_w,   H)),
        "bottom_right": rgb.crop((W-half_w, H-half_h,   W,  H))
    }

    return patches
```

```python
# patch an example rgb image

image_path = "../datasets/histology/testB/P12_ROI_01_C10_T_rgb.png"
rgb = Image.open(image_path)

patch_dict = patch_rgb(rgb)


plt.figure(figsize=(15, 4))

plt.subplot(1, 5, 1)
plt.imshow(rgb, cmap='gray')
plt.title("Original")
plt.axis("off")

plt.subplot(1, 5, 2)
plt.imshow(patch_dict["top_left"], cmap='gray')
plt.title("Top left")
plt.axis("off")

plt.subplot(1, 5, 3)
plt.imshow(patch_dict["top_right"], cmap='gray')
plt.title("Top right")
plt.axis("off")

plt.subplot(1, 5, 4)
plt.imshow(patch_dict["bottom_left"], cmap='gray')
plt.title("Bottom left")
plt.axis("off")

plt.subplot(1, 5, 5)
plt.imshow(patch_dict["bottom_right"], cmap='gray')
plt.title("Bottom right")
plt.axis("off")

plt.tight_layout()
plt.show()
```
